Content Diversity: What genres and categories dominate the platform?

```python
# --- BUSINESS QUESTION 1: Content Diversity (Genres) ---BAR CHART
import ast

def safe_parse_genre(x):
    if isinstance(x, str) and x.startswith('['):
        try:
            res = ast.literal_eval(x)
            return res[0] if len(res) > 0 else "Unknown"
        except:
            return "Unknown"
    return str(x)

top_genres = titles['genres'].apply(safe_parse_genre)
top_genres = top_genres[top_genres != "Unknown"].value_counts().head(8)

plt.figure(figsize=(10, 4))
ax = sns.barplot(x=top_genres.index, y=top_genres.values, palette="Blues_r")
plt.title("Business Q1: Top Genres Dominating the Platform", fontsize=12, weight="bold")
plt.xlabel("Genre", fontsize=11)
plt.ylabel("Count", fontsize=11)
plt.xticks(rotation=45)

for p in ax.patches:
    ax.annotate(f"{int(p.get_height())}", (p.get_x() + p.get_width() / 2., p.get_height()),
                ha="center", va="bottom", xytext=(0, 3), textcoords="offset points")

sns.despine(top=True, right=True)
plt.tight_layout()
plt.show()
```

Observation: Drama and comedy heavily dominate the platform's catalog, accounting for the vast majority of listed titles. This indicates a strong consumer preference and catalog focus on mainstream entertainment genres.

Regional Availability: How does content distribution vary across different regions (production countries)?

```python
# --- BUSINESS QUESTION 2: Regional Availability (Donut Chart) ---
import ast

def safe_parse_country(x):
    if isinstance(x, str) and x.startswith('['):
        try:
            res = ast.literal_eval(x)
            return res[0] if len(res) > 0 else "Unknown"
        except:
            return "Unknown"
    return str(x)

top_countries = titles['production_countries'].apply(safe_parse_country)
top_countries = top_countries[(top_countries != "Unknown") & (top_countries != "")].value_counts().head(6)

plt.figure(figsize=(8, 5))
colors = sns.color_palette("Purples_r", len(top_countries))
plt.pie(top_countries.values, labels=top_countries.index, autopct='%1.1f%%', startangle=140, colors=colors, wedgeprops=dict(width=0.4, edgecolor='w'))
plt.title("Business Q2: Regional Availability (Top Production Countries Share)", fontsize=12, weight="bold")
plt.tight_layout()
plt.show()
```

Observation: A select few global markets drive the vast majority of content production. The vast distribution skew highlights heavily concentrated regional content sourcing rather than an even global spread.



Trends Over Time: How has Amazon Prime’s content library evolved across release years?

```python
# --- BUSINESS QUESTION 3: Trends Over Time (Release Year Evolution) ---
# Group by release_year and count the number of titles released each year
yearly_trends = titles['release_year'].value_counts().sort_index()

# Filter for recent years (e.g., from 2000 onwards) to keep the trend clean and readable
yearly_trends = yearly_trends[yearly_trends.index >= 2000]

plt.figure(figsize=(11, 4))
ax = sns.lineplot(x=yearly_trends.index, y=yearly_trends.values, marker="o", color="teal", linewidth=2.5)

plt.title("Business Q3: Content Release Trends Over Time", fontsize=12, weight="bold")
plt.xlabel("Release Year", fontsize=11)
plt.ylabel("Number of Titles Released", fontsize=11)

sns.despine(top=True, right=True)
plt.tight_layout()
plt.show()
```

Observation: Content production shows a steady, aggressive upward trajectory in recent years. This reflects the platform's accelerated efforts to expand its digital catalog and scale up streaming availability over time.

IMDb Ratings & Popularity: What are the relationships between IMDb ratings and overall popularity (TMDB popularity) on the platform?

```python
# --- BUSINESS QUESTION 4: IMDb Ratings & Popularity (Scatter Plot) ---
# Drop missing values for clean correlation plotting
ratings_df = titles.dropna(subset=['imdb_score', 'tmdb_popularity'])

plt.figure(figsize=(9, 5))
ax = sns.scatterplot(
    data=ratings_df,
    x='imdb_score',
    y='tmdb_popularity',
    alpha=0.6,
    color="coral"
)

plt.title("Business Q4: IMDb Scores vs. TMDB Popularity", fontsize=12, weight="bold")
plt.xlabel("IMDb Score", fontsize=11)
plt.ylabel("TMDB Popularity", fontsize=11)

sns.despine(top=True, right=True)
plt.tight_layout()
plt.show()
```

Observation: While a large volume of content clusters around moderate scores, high IMDb critical ratings do not automatically guarantee peak TMDB popularity scores. Viewer engagement operates independently of critical acclaim.

Content Duration: What is the typical runtime distribution of movies on the platform, and how concentrated are standard feature-length runtimes?

```python
# --- BUSINESS QUESTION / CHART 5: Movie Runtime Distribution (Histogram & KDE) ---
# Filter out missing runtimes and focus on realistic movie lengths
runtimes = titles['runtime'].dropna()
runtimes = runtimes[(runtimes > 0) & (runtimes <= 250)]

plt.figure(figsize=(10, 4))
ax = sns.histplot(runtimes, kde=True, color="forestgreen", bins=30)

plt.title("Chart 5: Distribution of Movie Runtimes", fontsize=12, weight="bold")
plt.xlabel("Runtime (Minutes)", fontsize=11)
plt.ylabel("Density / Count", fontsize=11)

sns.despine(top=True, right=True)
plt.tight_layout()
plt.show()
```

Observation: Movie runtimes cluster tightly within the standard feature-length window of 90 to 120 minutes. Extreme short films or ultra-long epics represent rare exceptions in the library.

## Conclusion
This Exploratory Data Analysis of the Amazon Prime Video dataset successfully uncovered key patterns in platform content distribution, regional availability, historical publishing growth, audience metrics, and content runtimes.

* **Genre Dominance:** Drama leads the platform by a massive margin (2,626 titles), followed closely by comedy (1,608 titles) and thriller (1,101 titles), forming the heavy backbone of the streaming library's inventory[cite: 1].
* **Geographical Concentration:** Content production is overwhelmingly concentrated in the **United States (US)** at **65.7%**, followed by **India (IN)** at **13.9%** and the **United Kingdom (GB)** at **10.2%**, highlighting an execution strategy anchored by major Western and international media hubs[cite: 2].
* **Historical Release Trends:** Content output remained relatively low and stable through the 2000s before experiencing a sharp, aggressive upward surge starting in the early 2010s and peaking past 800 titles per year near 2021, reflecting the platform's rapid digital scaling[cite: 3].
* **Ratings vs. Popularity:** High IMDb critical scores do not guarantee peak TMDB popularity. The vast majority of titles cluster densely at moderate popularity levels across a wide spectrum of ratings, showing that mainstream audience engagement operates independently of high critical acclaim[cite: 4].
* **Runtime Distribution:** Movie runtimes cluster heavily around the standard feature-length window of **80 to 110 minutes** (peaking sharply near 90 minutes), demonstrating a clear industry preference for traditional theatrical runtime formats[cite: 5].

These insights paint a comprehensive picture of a US-centric, drama-focused catalog that expanded rapidly in the 2010s, while providing a solid, data-driven foundation for future content acquisition, recommendations, and platform planning.
